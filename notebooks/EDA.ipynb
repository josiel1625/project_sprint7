```python
import pandas as pd
import plotly.graph_objects as go # Importación de plotly.graph_objects como go

# Leer los datos del archivo CSV
car_data = pd.read_csv(r"C:\Users\josie\Documents\vehicles_env\project_sprint7\vehicles_us.csv")

# Crear un scatter plot utilizando plotly.graph_objects
# Se crea una figura vacía y luego se añade un rastro de scatter
fig = go.Figure(data=[go.Scatter(x=car_data['odometer'], y=car_data['price'], mode='markers')])

# Opcional: Puedes añadir un título al gráfico si lo deseas
fig.update_layout(title_text='Relación entre Odómetro y Precio')

# Mostrar el gráfico Plotly
fig.show()
```

```python
car_data['price'] = car_data['price'].astype(float)

car_data.info()
```

```python
car_data['model_year'] = car_data['model_year'].fillna(0).astype('int64')

car_data.info()
```

```python
import numpy as np
car_data['model_year'] = car_data['model_year'].replace(0, np.nan).astype('Int64')

car_data.info()
```

```python
faltantes = pd.DataFrame({
    'nulos': car_data.isna().sum(),
    'porcentaje': (car_data.isna().mean() * 100).round(1)
}).sort_values('porcentaje', ascending=False)
print(faltantes)

print('Duplicados exactos:', car_data.duplicated().sum())
```

```python
car_data.describe().T
car_data.describe(include=['object', 'string', 'category']).T
for col in ['condition', 'fuel', 'transmission', 'type', 'paint_color']:
    print(car_data[col].value_counts(dropna=False), '\n')
    
```

```python
car_data['type'].unique()
```

```python
# Para conocer la distribución de los tipos de vehículos, podemos utilizar un gráfico de barras:
car_data['type'].value_counts(dropna=False).plot(kind='bar', title='Distribución de Tipos de Vehículos')
```

```python
# Se grafica la distribución de la condición de los vehículos dentro de cada tipo utilizando un gráfico de barras apiladas. Se utiliza la función crosstab para calcular la proporción de cada condición dentro de cada tipo:

import matplotlib.pyplot as plt

tabla_pct = pd.crosstab(car_data['type'], car_data['condition'], normalize='index') * 100

tabla_pct.plot(kind='bar', stacked=True, figsize=(12, 5))
plt.title('Distribución de condición dentro de cada tipo (%)')
plt.ylabel('Porcentaje')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Condición', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Se van a analizar las distribuciones de las variables numéricas 'price', 'model_year', 'odometer', 'cylinders' y 'days_listed' utilizando histogramas.
num_cols = ['price', 'model_year', 'odometer', 'cylinders', 'days_listed']
car_data[num_cols].hist(bins=40, figsize=(14, 8))
plt.tight_layout(); plt.show()
```

```python

import seaborn as sns

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, ['price', 'odometer', 'days_listed']):
    sns.boxplot(x=car_data[col], ax=ax)
plt.show()
```
